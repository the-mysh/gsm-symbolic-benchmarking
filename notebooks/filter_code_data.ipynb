```python
import pandas as pd
import numpy as np
import os

from results_notebook_setup import RESULTS_FOLDERS, RESULTS_ROOT
```

```python
data_path = RESULTS_ROOT / RESULTS_FOLDERS['long_code']
data_path
```

```python
filtered_path = data_path.parent.parent / f"{data_path.parts[-2]}__filtered" / data_path.parts[-1]
filtered_path
```

```python
os.makedirs(filtered_path, exist_ok=True)
```

```python
for dset_folder in os.listdir(data_path):
    os.makedirs(filtered_path/dset_folder, exist_ok=True)
    for model_file in os.listdir(data_path/dset_folder):
        if model_file.endswith('.parquet'):
            df = pd.read_parquet(data_path/dset_folder/model_file)
            df = df[~np.isnan(df.predicted_numerical_result)]
            df.to_parquet(filtered_path/dset_folder/model_file)
```
